```python
# [조건 1] 이름, 학과, 관심 분야 출력
# [조건 2] \n (줄바꿈 제어문자) 사용
# [조건 3] end (끝 문자 제어) 사용
# [조건 4] f-string을 활용한 출력
# [조건 5] 변수 생성 및 정보 입력
```

```python
name = "양지성"
age = 26
department = "정보통신공학과"
interests = "클라우드 & 네트워크 인프라, 인공지능"
subject = "Python & 머신러닝"


print("========", "자기소개", "========", sep=" ")
print("이름", name, sep="\t\t: ")
print("학과", department, sep="\t\t: ")
print("관심 분야", interests, sep="\t: ")


print("\n\n안녕하세요!!", end=" ")
print(f"제 이름은 {name}이고 나이는 {age}살입니다.")
print(f"{subject} 과목을 열심히 공부하겠습니다!")
```
